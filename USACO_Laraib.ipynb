# Crtanje

```python
# Read input
n = int(input())
s = input().strip()

# Current net worth
value = 0

# Store drawn characters with their (row, column)
points = []

min_row = 0
max_row = 0

for col in range(n):
    if s[col] == '+':
        # draw '/' at value before increase
        row = value
        points.append((row, col, '/'))
        value += 1

    elif s[col] == '-':
        # decrease first, then draw '\'
        value -= 1
        row = value
        points.append((row, col, '\\'))

    else:  # '='
        row = value
        points.append((row, col, '_'))

    min_row = min(min_row, row)
    max_row = max(max_row, row)

# Height of matrix
height = max_row - min_row + 1

# Create matrix filled with dots
matrix = []
for _ in range(height):
    matrix.append(['.'] * n)

# Place characters
for row, col, ch in points:
    # Convert real row to matrix index
    matrix[max_row - row][col] = ch

# Print result
for row in matrix:
    print(''.join(row))
```

```python
# Rue's Rings - Easy Python Solution

for _ in range(10):   # 10 datasets
    n = int(input())

    smallest = 10**9
    routes = []

    for i in range(n):
        data = list(map(int, input().split()))

        route_id = data[0]
        k = data[1]
        diameters = data[2:]

        mini = min(diameters)

        # if new smallest found
        if mini < smallest:
            smallest = mini
            routes = [route_id]

        # if same smallest found
        elif mini == smallest:
            routes.append(route_id)

    routes.sort()

    # convert list to required format
    ans = ",".join(map(str, routes))

    print(f"{smallest} {{{ans}}}")
```

# Terrific Triangles


```python
# number of questions
n = int(input())

for i in range(n):
    # read the three sides
    a, b, c = map(int, input().split())

    # sort the sides so the largest is last
    sides = sorted([a, b, c])

    a = sides[0]
    b = sides[1]
    c = sides[2]

    # check the triangle type
    if a*a + b*b == c*c:
        print("R")   # Right triangle
    elif a*a + b*b > c*c:
        print("A")   # Acute triangle
    else:
        print("O")   # Obtuse triangle
```

# Teleportation

```python
# Input: four integers
# a = starting location (where manure is now)
# b = destination location (where manure must go)
# x, y = teleporter endpoints (you can travel instantly between them)
a, b, x, y = map(int, input().split())


# Distances between starting/ending points and teleporter endpoints
a_x = abs(a - x)  # distance from a to teleporter at x
a_y = abs(a - y)  # distance from a to teleporter at y
b_x = abs(b - x)  # distance from b to teleporter at x
b_y = abs(b - y)  # distance from b to teleporter at y


# Case: using the teleporter
# Two possible strategies:
# 1. Haul manure from a → x, teleport x → y, then haul y → b
# 2. Haul manure from a → y, teleport y → x, then haul x → b
# We compute whichever of these is shorter
farmer_dist = 0
if a_x < a_y:
    farmer_dist += a_x   # choosing to go from a to x
    farmer_dist += b_y   # no other option : haul y → b
else: #a_x > a_y or a_x = a_y
    farmer_dist += a_y   # # choosing to go from a to x
    farmer_dist += b_x   # no other option : haul y → b


# Case: not using teleporter (direct hauling only)
direct_distance = abs(a - b)


# Final answer: the minimum of (direct hauling) vs (using teleporter)
print(min(direct_distance, farmer_dist))
```

# Christmas

```python
t = int(input())

for _ in range(t):
    n, d, h = map(int, input().split())
    y = list(map(int, input().split()))

    # total area = sab triangles ka area
    total_area = n * (d * h / 2)

    # har do adjacent triangles check karo
    for i in range(1, n):
        # previous triangle ka top
        prev_top = y[i - 1] + h

        # agar overlap ho raha hai
        if y[i] < prev_top:
            overlap_height = prev_top - y[i]

            # chhote overlap triangle ka base (similar triangles)    (small base/big base = small height/big height)
            overlap_base = d * overlap_height / h

            # chhote triangle ka area
            overlap_area = overlap_base * overlap_height / 2

            # extra area minus
            total_area -= overlap_area

    print(total_area)
```

# Milk measurement

```python
n, x = map(int, input().split())
weights = list(map(int, input().split()))

weights.sort()

left = 0
right = n - 1
gondolas = 0

while left <= right:
    if weights[left] + weights[right] <= x:
        left += 1
        right -= 1
    else:
        right -= 1
    gondolas += 1

print(gondolas)
```

```python
buc = int(input())                 # Number of buckets
cap = list(map(int, input().split()))  # Capacities
wat = list(map(int, input().split()))  # Water in buckets

# Pour water from bucket i into bucket i+1
for i in range(buc-1):
    giver = wat[i]          # current bucket
    receiver = wat[i + 1]   # next bucket
    rec_cap = cap[i + 1]

    rem_cap = rec_cap - receiver

    if giver <= rem_cap:
        receiver += giver
        giver = 0
    else:
        receiver = rec_cap
        giver -= rem_cap
    wat[i]=giver
    wat[i + 1]= receiver

print(*wat)
```

# Shell Game

```python
# Initial arrangement of shells
shells = ['yellow', 'pink', 'blue']


num_swaps = int(input("Enter number of swaps: "))


# Read all swaps using a for loop bcz there maybe more than 3 swaps
# converts each swap into list of format [a, b, g] where:
# a = first position to swap, b = second position to swap, g = guess
swaps = []
for _ in range(num_swaps):
    swaps.append([int(x) - 1 for x in input().split()])

#doing -1 to convert to 0-based indexing


# Print the swap list (for debugging, optional)
#print("Swaps entered:", swaps)


score_yellow = 0 #scores when stone is under yellow shell
score_pink = 0 #scores when stone is under pink shell
score_blue = 0 #scores when stone is under blue shell


for a, b, g in swaps: # unpack each swap
    # Swap index a and index b in the shells list
    shells[a], shells[b] = shells[b], shells[a]


    # Check if Elsie's guess is correct and update scores
    if shells[g] == 'yellow':
        score_yellow += 1
    elif shells[g] == 'pink':
        score_pink += 1
    else:  # must be orange
        score_blue += 1


# Print the maximum score (the best possible outcome)
print(max(score_yellow, score_pink, score_blue))
```

# Teleportation

```python
# Input: four integers
# a = starting location (where manure is now)
# b = destination location (where manure must go)
# x, y = teleporter endpoints (you can travel instantly between them)
a, b, x, y = map(int, input().split())


# Distances between starting/ending points and teleporter endpoints
a_x = abs(a - x)  # distance from a to teleporter at x
a_y = abs(a - y)  # distance from a to teleporter at y
b_x = abs(b - x)  # distance from b to teleporter at x
b_y = abs(b - y)  # distance from b to teleporter at y


# Case: using the teleporter
# Two possible strategies:
# 1. Haul manure from a → x, teleport x → y, then haul y → b
# 2. Haul manure from a → y, teleport y → x, then haul x → b
# We compute whichever of these is shorter
farmer_dist = 0
if a_x < a_y:
    farmer_dist += a_x   # choosing to go from a to x
    farmer_dist += b_y   # no other option : haul y → b
else: #a_x > a_y or a_x = a_y
    farmer_dist += a_y   # # choosing to go from a to x
    farmer_dist += b_x   # no other option : haul y → b


# Case: not using teleporter (direct hauling only)
direct_distance = abs(a - b)


# Final answer: the minimum of (direct hauling) vs (using teleporter)
print(min(direct_distance, farmer_dist))
```

```python
f = int(input())               # Take input N → how many mailboxes/farms are there
s = input().strip()            # Take the mailbox colors string; .strip() removes extra spaces/newlines

k = 1                          # Start checking with the smallest window size (K = 1)

while k <= f:                  # Try K from 1 up to N, until we find the answer
    patterns = []              # This list will store all substrings of length K (to check duplicates)
    for i in range(f - k + 1): # Loop over all possible starting positions; total (N-K+1) substrings
        st = s[i:i+k]          # Cut out the current substring: start at i, take length K
        if st in patterns:     # If this substring was already seen → duplicate found
            break              # Stop checking for this K, it's not valid
        patterns.append(st)    # Otherwise, add this new substring to the list

    else:                      # FOR-ELSE: runs only if loop finishes without a 'break'
        print(k)               # Means all substrings were unique → we found the smallest valid K
        break                  # Print answer and stop program

    k += 1                     # If we found a duplicate, increase K and try again
```

```python
x, y = map(int, input().split())

total = 0
og_x = x
step = 1

while True:
    # Farmer John x se step ke hisaab se naya point jaane ki koshish karega
    new_pos = og_x + step

    # Agar y (cow) og_x aur new_pos ke beech ho jaye (inclusive), to
    if (og_x <= y <= new_pos) or (new_pos <= y <= og_x):
        # Sirf utni distance add karo jitni og_x se y tak hai
        total += abs(y - og_x)
        break
    else:
        # Cow beech mein na ho, to Farmer John og_x → new_pos jaata hai, phir wapas og_x
        # Us poore round ka distance: og_x → new_pos aur wapas new_pos → og_x
        total += abs(new_pos - og_x)*2

    # Farmer John wapas og_x pe aata hai (strategy ki wajah se)
    # Is liye ham og_x ko badalne ki zaroorat **nahin** — wo wahi rahega

    # Next step me direction aur distance dono double hone chahiye
    step *= -2

print(total)
```

```python
# Distance² = (x1 - x2)² + (y1 - y2)²

# Example input (we can imagine these as 3 points on a map)
n = 3
x = [321, -15, -525]   # x-coordinates of each point
y = [404, 373, 990]    # y-coordinates of each point


max_dist = 0


# Go through every pair of points to find which two are farthest apart
for i in range(n):
    for j in range(n):
        # Step 1: apply formula
        dist = (x[i] - x[j])**2 + (y[i] - y[j])**2


        # Step 2: Keep track of the largest distance found so far
        max_dist = max(max_dist, dist)


# After checking all pairs, print the largest squared distance
print(max_dist)
```

```python
# Read N and Q
N, Q = map(int, input().split())

# Read costs of each deal
cost = list(map(int, input().split()))

# Preprocess: make sure buying two smaller deals is not more expensive than bigger deals
for i in range(1, N):
    if cost[i] > 2 * cost[i-1]:
        cost[i] = 2 * cost[i-1]

# Process each query
for _ in range(Q):
    z = int(input())
    total = 0
    remaining = z
    res = float('inf')

    # Greedily take deals from largest to smallest
    for i in range(N-1, -1, -1):
        buckets = 1 << i  # 2^i buckets
        take = remaining // buckets
        total += take * cost[i]
        remaining -= take * buckets

        # Consider buying one extra of this deal to cover remaining
        if remaining > 0:
            res = min(res, total + cost[i])

    # Check if total without extra is better
    res = min(res, total)

    print(res)
```
